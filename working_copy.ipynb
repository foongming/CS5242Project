```python
def get_todays_news(soup) -> list[dict | None]:
    news_trio = soup.find('div', class_='news-trio')

    if not news_trio:
        return []

    news_items = news_trio.find_all('div', class_='news-item')

    if not news_items:
        return []

    d = []
    for i, news in enumerate(news_items):
        news_struct = {}
        a_tag = news.find('a')
        news_struct['title'] = a_tag.get_text(strip=True)
        news_struct['allitems-link'] = a_tag.get('href')
        d.append(news_struct)
    return d

html = driver.page_source
soup = BeautifulSoup(html, 'html.parser')
get_todays_news(soup)
```
